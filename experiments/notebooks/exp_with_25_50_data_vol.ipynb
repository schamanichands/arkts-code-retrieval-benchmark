```python
from datasets import load_dataset

dataset = load_dataset("hreyulog/arkts-code-docstring")

print(dataset)
```

```python
MODEL_NAME = "google/embeddinggemma-300m"
BATCH_SIZE = 4
EPOCHS = 2
LEARNING_RATE = 1e-5
MAX_SEQ_LENGTH = 512
WARMUP_RATIO = 0.10
SEED = 42
USE_AMP = False
```

```python
model = SentenceTransformer(
    MODEL_NAME,
    device="cuda"
)
```

```python
model.max_seq_length = MAX_SEQ_LENGTH

print("Model loaded:", MODEL_NAME)
print("Max sequence length:", model.max_seq_length)
print("Device:", model.device)
```

```python
train_dataset = dataset["train"]
from torch.utils.data import DataLoader
```

```python
# =========================
# Prepare training examples
# =========================

train_examples = [
    InputExample(
        texts=[
            row["docstring"],
            row["function"]
        ]
    )
    for row in train_dataset
]

train_dataloader = DataLoader(
    train_examples,
    shuffle=True,
    batch_size=BATCH_SIZE,
    collate_fn=model.smart_batching_collate
)
```

```python
OUTPUT_DIR = "/content/drive/MyDrive/arkts_exp/embeddinggemma-300m-arkts-100-safe"
```

```python
import os
import json
import torch

# =========================
# Loss
# =========================

train_loss = losses.MultipleNegativesRankingLoss(model)

# =========================
# Training parameters
# =========================

steps_per_epoch = len(train_dataloader)
total_steps = steps_per_epoch * EPOCHS
warmup_steps = int(total_steps * WARMUP_RATIO)

print("\nTraining configuration:")
print("Train size:", len(train_examples))
print("Batch size:", BATCH_SIZE)
print("Epochs:", EPOCHS)
print("Learning rate:", LEARNING_RATE)
print("Max sequence length:", MAX_SEQ_LENGTH)
print("Warmup steps:", warmup_steps)
print("AMP:", USE_AMP)
print("Total steps:", total_steps)

# =========================
# Save experiment config
# =========================

config = {
    "model": MODEL_NAME,
    "dataset": "hreyulog/arkts-code-docstring",
    "train_size": len(train_examples),
    "train_fraction": 1.0,
    "batch_size": BATCH_SIZE,
    "epochs": EPOCHS,
    "learning_rate": LEARNING_RATE,
    "max_seq_length": MAX_SEQ_LENGTH,
    "loss": "MultipleNegativesRankingLoss",
    "warmup_ratio": WARMUP_RATIO,
    "warmup_steps": warmup_steps,
    "use_amp": USE_AMP,
    "seed": SEED,
    "output_dir": OUTPUT_DIR
}

with open(
    os.path.join(OUTPUT_DIR, "experiment_config.json"),
    "w"
) as f:
    json.dump(config, f, indent=2)
```

```python
# =========================
# Train
# =========================

torch.manual_seed(SEED)

model.fit(
    train_objectives=[
        (train_dataloader, train_loss)
    ],
    epochs=EPOCHS,
    warmup_steps=warmup_steps,
    optimizer_params={
        "lr": LEARNING_RATE
    },
    output_path=OUTPUT_DIR,
    show_progress_bar=True,
    use_amp=USE_AMP
)

print("\nTraining finished.")
print("Model saved to:")
print(OUTPUT_DIR)
```

```python
from sentence_transformers import SentenceTransformer

model_path = "/content/drive/MyDrive/arkts_exp/embeddinggemma-300m-arkts-100-safe"

model = SentenceTransformer(model_path, device="cuda")
model.max_seq_length = 512

print("Model loaded")
```

```python
from datasets import load_dataset

dataset = load_dataset("hreyulog/arkts-code-docstring")

test_ds = dataset["test"]

print(len(test_ds))
```

```python
import numpy as np

docstrings = test_ds["docstring"]
functions = test_ds["function"]

query_embeddings = model.encode(
    docstrings,
    batch_size=32,
    normalize_embeddings=True,
    show_progress_bar=True,
)

code_embeddings = model.encode(
    functions,
    batch_size=32,
    normalize_embeddings=True,
    show_progress_bar=True,
)

print(query_embeddings.shape)
print(code_embeddings.shape)
```

```python
similarities = query_embeddings @ code_embeddings.T

print(similarities.shape)
```

```python
ranks = []

for i in range(len(test_ds)):
    scores = similarities[i]

    # Чем больше similarity, тем выше место
    rank = 1 + np.sum(scores > scores[i])

    ranks.append(rank)

ranks = np.array(ranks)

print("Mean rank:", ranks.mean())
print("Median rank:", np.median(ranks))
```

```python
recall_at_1 = np.mean(ranks <= 1)
recall_at_5 = np.mean(ranks <= 5)

mrr = np.mean(1.0 / ranks)

ndcg_at_5 = np.mean(
    np.where(ranks <= 5, 1.0 / np.log2(ranks + 1), 0.0)
)

print(f"R@1:    {recall_at_1:.4f}")
print(f"R@5:    {recall_at_5:.4f}")
print(f"MRR:    {mrr:.4f}")
print(f"NDCG@5: {ndcg_at_5:.4f}")
```

```python
from datasets import load_dataset
from sentence_transformers import SentenceTransformer, InputExample, losses
from torch.utils.data import DataLoader
import math
```

```python
model = SentenceTransformer(
    "google/embeddinggemma-300m",
    device="cuda"
)
model.max_seq_length = 512
```

```python
dataset = load_dataset("hreyulog/arkts-code-docstring")

train_full = dataset["train"]

train_50 = train_full.shuffle(seed=42).select(
    range(int(len(train_full) * 0.50))
)

print("Train examples:", len(train_50))
```

```python
train_examples = [
    InputExample(
        texts=[row["docstring"], row["function"]]
    )
    for row in train_50
]

train_dataloader = DataLoader(
    train_examples,
    shuffle=True,
    batch_size=4,
    collate_fn=model.smart_batching_collate
)

train_loss = losses.MultipleNegativesRankingLoss(model)
```

```python
epochs = 2
lr = 1e-5
warmup_steps = math.ceil(
    len(train_dataloader) * epochs * 0.10
)

print("Steps:", len(train_dataloader) * epochs)
print("Warmup steps:", warmup_steps)
```

```python
output_path = "/content/drive/MyDrive/arkts_exp/embeddinggemma-300m-arkts-50-safe"

model.fit(
    train_objectives=[(train_dataloader, train_loss)],
    epochs=2,
    warmup_steps=warmup_steps,
    optimizer_params={"lr": 1e-5},
    show_progress_bar=True,
    output_path=output_path,
    use_amp=False
)

print("Training finished.")
print(output_path)
```
